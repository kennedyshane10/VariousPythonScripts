```python
def first(word):
    """Returns the first character of a string."""
    return word[0]

def last(word):
    """Returns the last character of a string."""
    return word[-1]

def middle(word):
    """Returns all but the first and characters of a string."""
    return word[1:-1]

def is_palindrome(word):
    """Returns True if word is a palindrome."""
    if len(word) <= 1:
        return True
    if first(word) != last(word):
        return False
    else:
       return is_palindrome(middle(word))  
```

```python
def simulate(prog, xsize, ysize):
    """Given a "program" *prog*, run that program by "moving around" and
    tracking what happens. Use a set of tuples to track locations visited.

    Return the number of junctions visited and the number of
    time-steps used.
    
    Examples of usage and results:

    >>> simulate("NNN", 100, 100)
    (4, 3)
    >>> simulate("NNNSSS", 100, 100)
    (4, 6)
    >>> simulate("NNNESSS", 100, 100)
    (8, 7)
    >>> simulate("NNNNNSSSSS", 3, 3)
    (3, 10)
    >>> simulate(plan(10, 10), 10, 10)
    (100, 99)
    """

    #Tuple for tracking locations visited
    x, y = 0, 0
    visited = {(x,y)}

    #Movement directions
    directions = {
    'N': (0,1),
    'S': (0,-1),
    'E': (1,0),
    'W': (-1,0)
    }

    # Iterate over each step in the program
    for step in prog:
        dx, dy = directions[step]
        new_x, new_y = x + dx, y + dy

    # Check if the new position is within the grid limits
        if 0 <= new_x < xsize and 0 <= new_y < ysize:
            x, y = new_x, new_y
            visited.add((x, y))    

    # my len
    def len(var):
        l = 0
        for i in var:
            l += 1
        return l
    
    # Return the number of junctions visited and the number of time-steps used
    return len(visited), len(prog)    
```

```python
simulate("NNNNNSS",3,3)
```

```python
def plan(xsize, ysize):
    """Given the city limits, return a program which moves around the
    entire city. The program should be a string, eg "NNESS".

    By the way, notice that doctest will insist that the "expected"
    string, below, must be in single-quotes (''), not double-quotes
    ("").

    >>> plan(2, 2)
    'NES'
    >>> plan(2, 3)
    'NNESS'

    """
    prog = []
    
    # Traverse the grid row by row
    for y in range(ysize):
        if y % 2 == 0:
            # Move east in even rows
            prog.extend(['E'] * (xsize - 1))
        else:
            # Move west in odd rows
            prog.extend(['W'] * (xsize - 1))
        
        # Move north to the next row if not at the last row
        if y < ysize - 1:
            prog.append('N')
    
    return ''.join(prog)
```

```python
def plan_alternative(xsize, ysize):
    """Given the city limits, return a program which moves around the
    entire city. The program should be a string, eg "NNESS".

    By the way, notice that doctest will insist that the "expected"
    string, below, must be in single-quotes (''), not double-quotes
    ("").

    Our plan is: starting from the southwest, we traverse all the way
    northwards, then one step east, then all the way southwards, then
    one step east, and so on.

    >>> plan(2, 2)
    'NES'
    >>> plan(2, 3)
    'NNESS'

    """
    prog = []
    
    # Traverse the grid column by column
    for y in range(xsize):
        if y % 2 == 0:
            # Move east in even rows
            prog.extend(['N'] * (ysize - 1))
        else:
            # Move west in odd rows
            prog.extend(['S'] * (ysize - 1))
        
        # Move north to the next row if not at the last row
        if y < xsize - 1:
            prog.append('E')
    
    return ''.join(prog)
```

Write a function to calculate the cumulative sum of a list. It will also work on an input tuple, but will still return a list.

```python
def cumsum(L):
    """
    >>> cumsum([5, 5, 5])
    [5, 10, 15]
    >>> cumsum([])
    []
    >>> cumsum([5])
    [5]
    >>> cumsum((5, 5, 5))
    [5, 10, 15]
    """
    s = 0
    result = []
    for x in L:
        s += x
        result.append(s)
    return result
```

```python
doctest.run_docstring_examples(cumsum, globals(), verbose=True)
```

```python
def product(L):
    """
    >>> product([1, 1, 1])
    1
    >>> product([1, 2, 3])
    6
    >>> product([]) # the sum of an empty list is 0, but the product of an empty list is 1
    1
    """
    result = 1
    for x in L:
        result *= x
    return result
```

```python
def cuboids(n):
    # we *canonicalise* on the ordering x <= y <= z to avoid duplicates
    for x in range(1, n):
        for y in range(x, n):
            for z in range(y, n):
                yield (x, y, z)
```

```python
def n_cuboids_side_lt(side):
    """
    >>> n_cuboids_side_lt(250) # side_lt means "sides are less than"
    2604125
    """
    return sum(1 for c in cuboids(side))
```

```python
doctest.run_docstring_examples(n_cuboids_side_lt, globals(), verbose=True)
```
