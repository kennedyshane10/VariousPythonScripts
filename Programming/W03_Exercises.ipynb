**Exercise**: in `W03_01_Regression_by_Guessing.ipynb` we suggested to try to produce some better guesses for the function $y=f(x)$. What is the best function you can find? What RMSE and $R^2$ do you achieve? Use the code below to start.

```python
filename = "C:/Users/skenne26/OneDrive - UHG\Myfolder/12_MSc/CT5148_ProgrammingAndToolsForAI/W03/data/cell_growth.csv"
```

```python
df = pd.read_csv(filename)
```

```python
df.head()
```

```python
x = df['x']
y = df['y']
```

```python
plt.scatter(x,y)
plt.xlim((0,10))
plt.ylim((0,15))
plt.xlabel("x axis")
plt.ylabel("y axis");
```

```python
def f_guess1(x): return np.log(0.5*x**2 +1)
```

```python
def f_guess2(x): return x**2
```

```python
def f_guess3(x): return x**3
```

```python
def f_guess4(x): return np.exp(x)
```

```python
def f_guess5(x): return (np.exp(x) - 2*(np.exp(x/2)) + np.exp(x/3))
```

```python
guesses = [f_guess1, f_guess2, f_guess3, f_guess4, f_guess5]
```

```python
xgrid = np.linspace(0,10,101,endpoint=True)
```

```python
plt.scatter(x,y, label="Data")

for guess in guesses:
    plt.plot(xgrid,guess(xgrid), label=guess.__name__)

plt.xlim((0,10)); plt.ylim((0,15))
plt.xlabel("x axis"); plt.ylabel("y axis")
plt.legend();
```
